# CNN transfer learning for SonicSentinel
Uses Google YAMNet as a frozen audio CNN and trains a classifier for all ten SonicSentinel classes. Source recordings are split before segmentation. Validation macro-F1 selects the classifier; test data is evaluated afterward.
Place the updated code and audio_dataset/raw/<class>/ recordings in Drive. The pretrained CNN is downloaded once. Existing classical model artifacts are preserved.


```python
from google.colab import drive
from pathlib import Path
import subprocess
import sys
import os

drive.mount('/content/drive')
REPO_DIR = Path('/content/drive/MyDrive/SonicSentinelAI/Sonic-sentinal AI')
if not (REPO_DIR / 'requirements.txt').is_file():
    raise FileNotFoundError('Set REPO_DIR to the repository folder in Drive.')
os.chdir(REPO_DIR)
subprocess.run([sys.executable, '-m', 'pip', 'install', '-r', 'requirements.txt'], check=True)
```

```python
# Cached features allow an interrupted extraction to resume.
# --activate selects the trained model for the Flask app and backs up the old selection.
subprocess.run([sys.executable, '-u', 'python_models/train_transfer.py', '--extract', '--activate'], check=True)
```

```python
import json
metrics = json.loads(Path('python_models/yamnet_transfer_metrics.json').read_text())
print('Selected:', metrics['candidate'])
print('Validation accuracy:', metrics['validation']['accuracy'])
print('Test accuracy:', metrics['test']['accuracy'])
print('Test macro-F1:', metrics['test']['macro_f1'])
print('Split counts:', metrics['split_counts'])
```
